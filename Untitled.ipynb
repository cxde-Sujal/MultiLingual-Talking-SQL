````python
chain=create_sql_query_chain(llm,db)
def execute_query(question):
    try:
        #generate sql query from question
        response= chain.invoke({"question":question})
        print(response)
        print("--------------------------------")
        #clean the sql query to main sql part
        cleaned_query= response.strip('```sql\n').strip('\n```')
        print(cleaned_query)
        print("--------------------------------")
        #run the sql query
        result=db.run(cleaned_query)
        #display result
        print(result)
        # print("--------------------------------")
    except ProgrammingError as e:
        print(f"An error occured as {e}")

    
        
        
````

````python
q1="How many unique customers are there for each product category"
execute_query(q1)
````

````python
q2="Calculate total sales amount for each product category"
execute_query(q2)
````
